```python
import pandas as pd
import numpy as np

print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
print("Kernel is working!")
```

```python
from pathlib import Path

raw_folder = Path("../data/raw")

files = list(raw_folder.glob("*.csv"))

print("CSV files found:")
for file in files:
    print(file)

csv_file = files[0]

df = pd.read_csv(csv_file)

print(f"\nLoaded: {csv_file.name}")
print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]:,}")

df.head()
```

```python
# Basic dataset audit

print("Shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isna().sum().sort_values(ascending=False))

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
# Basic numerical/statistical summary

df.describe(include="all").T
```

```python
# Check data types and blank values

print("DATA TYPES")
print(df.dtypes)

print("\nBLANK STRINGS")
for col in df.columns:
    blanks = (df[col].astype(str).str.strip() == "").sum()
    if blanks > 0:
        print(f"{col}: {blanks} blank values")

print("\nNaN VALUES")
print(df.isna().sum().sort_values(ascending=False))
```

```python
# Inspect TotalCharges specifically

print("TotalCharges data type:", df["TotalCharges"].dtype)

print("\nBlank TotalCharges rows:")
print(df[df["TotalCharges"].astype(str).str.strip() == ""][["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]])
```

```python
# Convert TotalCharges from text to numeric.
# Blank strings become NaN.

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

print("TotalCharges data type:")
print(df["TotalCharges"].dtype)

print("\nMissing TotalCharges after conversion:")
print(df["TotalCharges"].isna().sum())
```

```python
missing_total = df[df["TotalCharges"].isna()]

print(missing_total[
    ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]
])
```

```python
# Fill missing TotalCharges for customers with zero tenure.
# These represent new customers with no accumulated charges.

mask = (df["TotalCharges"].isna()) & (df["tenure"] == 0)

print("Rows to fill:", mask.sum())

df.loc[mask, "TotalCharges"] = 0

print("Remaining missing TotalCharges:",
      df["TotalCharges"].isna().sum())
```

```python
df.loc[mask, ["customerID", "tenure", "MonthlyCharges", "TotalCharges", "Churn"]]
```

```python
# Check for duplicate records and duplicate customer IDs

print("Duplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:", df["customerID"].duplicated().sum())

print("\nNumber of unique customers:", df["customerID"].nunique())
print("Number of rows:", len(df))
```

```python
# Check the Churn target

print(df["Churn"].value_counts())

print("\nChurn percentage:")
print(df["Churn"].value_counts(normalize=True).mul(100).round(2))
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:", df["customerID"].duplicated().sum())
print("Unique customers:", df["customerID"].nunique())
print("Total rows:", len(df))
```

```python
# Inspect unique values in every categorical column

categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    print(f"\n{'=' * 50}")
    print(f"{col}")
    print(f"{'=' * 50}")
    print(df[col].value_counts(dropna=False))
```

```python
# Create a cleaned copy of the dataset

df_clean = df.copy()

# Remove accidental leading/trailing whitespace from string columns
string_cols = df_clean.select_dtypes(include=["str", "object"]).columns

for col in string_cols:
    df_clean[col] = df_clean[col].str.strip()

# Verify the final structure
print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))
print("Missing values:", df_clean.isna().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
from pathlib import Path

processed_folder = Path("../data/processed")
processed_folder.mkdir(parents=True, exist_ok=True)

output_file = processed_folder / "telco_customer_churn_clean.csv"

df_clean.to_csv(output_file, index=False)

print(f"Saved cleaned dataset to: {output_file}")
```
